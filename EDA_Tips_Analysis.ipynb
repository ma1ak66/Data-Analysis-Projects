# Exploratory Data Analysis (EDA) on Tips Dataset

## 1. Setup & Data Loading
Import necessary libraries (`pandas`, `matplotlib`, `seaborn`) and load the builtin `tips` dataset.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
tips = sns.load_dataset("tips")
tips.head()
```

## 2. Scatter Plot: Total Bill vs. Tip

**Goal:** Analyze the relationship between the total bill amount and the tip given, categorized by the time of day (Dinner vs. Lunch).

```python
# Create scatter plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")

plt.title("Total Bill vs. Tip by Time of Day")
plt.xlabel("Total Bill ($)")
plt.ylabel("Tip ($)")

plt.show()
```

## 3. Bar Plot: Average Tip by Day

**Goal:** Compare the average tip amount received across different days of the week.

```python
# Create bar plot for average tip per day
plt.figure(figsize=(8, 5))
sns.barplot(data=tips, x="day", y="tip", errorbar=None)

plt.title("Average Tip by Day of the Week")
plt.xlabel("Day of the Week")
plt.ylabel("Average Tip ($)")

plt.show()
```

## 4. Box Plot: Tip Distribution for Smokers vs. Non-Smokers

**Goal:** Visualize the spread, median, and potential outliers of tips given by smokers versus non-smokers.

```python
# Create box plot
plt.figure(figsize=(8, 5))
sns.boxplot(data=tips, x="smoker", y="tip")

plt.title("Tip Distribution: Smoker vs. Non-Smoker")
plt.xlabel("Smoker Status")
plt.ylabel("Tip ($)")

plt.show()
```

## 5. Correlation Heatmap

**Goal:** Examine numerical correlations between total bill, tip amount, and party size.

```python
# Compute correlation matrix for numerical columns
corr_matrix = tips.select_dtypes(include=["float64", "int64"]).corr()

# Create correlation heatmap
plt.figure(figsize=(6, 4))
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Numerical Columns")

plt.show()
```
